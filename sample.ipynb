# Simulator

Note: All simulation dynamics are defined in the single `mums.py` script.

* Use `EnvMaker` to create sample environments

```python
from mums import EnvMaker
```

* List available pre-built worlds

```python
available = EnvMaker.ListDataBase()
print(available)
```

* Create a pre-built world (this is compatible with `gymnasium`)

```python
env = EnvMaker.MakeFromDataBase(
    discrete=True,  # action space
    name="world1",  # name of the world
    freeze=True,    # if True, uses same seed on reset
    seed=None,      # seed for RNG
)
```

* Simulate `gymnasium` style env

```python
# reset the env to get initial state

obs, info = env.reset()
print(f"""
{obs.keys()} 

{obs['stations'] = }

{obs['users'] = }

{obs['association'] = }
""")
env.render()
```

```python
# step through the environment

done = False

while not done:
    action = env.action_space.sample() #<--- random action

    obs, reward, term, trunc, info = env.step(action)
    done = term or trunc
    
    print(f"""
    {action = }

    {reward = }

    {term = }, {trunc = }

    {obs['stations'] = }

    {obs['users'] = }

    {obs['association'] = }
    """)
    env.render()

    
```

# Compare with some Baselines

```python
from mums import EnvMaker, ContinuousBaselines, DiscreteBaselines
```

```python
env = EnvMaker.MakeFromDataBase(
    discrete=True,  # action space
    name="world1",  # name of the world
    freeze=False,    # if True, uses same seed on reset
    seed=None,      # seed for RNG
)
```

```python
pie = DiscreteBaselines(env, seed=None) if env.discrete else ContinuousBaselines(env, seed=None)
hs = []
for p in pie:
    res = pie(
        p,
        episodes=10,
        gamma=1.0,
        render=False,
        verbose=False,
    )
    print(p, res['mean_return'])
    hs.append((res['mean_return'],p))
sorted(hs)
```
